### Estimated number of rubin alerts compared to 


Goven the magnitude limits of both telescopes, the sensitive flux difference between them is:
$$
M_{ZTF} - M_{VRO} = -2.5\,log_{10}\frac{F_{VRO}}{F_{ZTF}}
$$

From which:
$$
\frac{F_{VRO}}{F_{ZTF}} = 10^{(M_{ZTF}-M_{VRO})/-2.5}
$$

Then, since $F\propto d^2$:
$$
\frac{d_\text{VRO}}{d_\text{ZTF}} = \sqrt{\frac{F_\text{VRO}}{F_\text{ZTF}}}
$$

Then, the number of sources will be proportional to $N\propto \rho \cdot V$. If we assume, to first order, that the density of sources is constant, then since $V\propto \frac{4\pi d^3}{3}$

$$
\frac{N_\text{VRO}}{N_\text{ZTF}} = \bigg(\frac{d_\text{VRO}}{d_\text{ZTF}}\bigg)^3 = \bigg(\frac{F_\text{VRO}}{F_\text{ZTF}}\bigg)^{3/2}
$$

```python
import numpy as np
```

```python
def get_flux_factor(M_ztf, M_vro):
    return 10 ** ((M_ztf - M_vro) / -2.5)

def get_N_factor(M_ztf, M_vro):
    flux_factor = get_flux_factor(M_ztf, M_vro)


    return flux_factor ** (3/2)
```
