```python
import json
import sys
```

```python
with open('../simulated_alerts.txt', 'r') as file:
    lines = file.readlines()
```

```python
alerts = []
LSST_alerts = []
ZTF_alerts = []

LSST_names = []
ZTF_names = [] #so we don't load in repeated ones
for line in lines:
    line.strip()
    alert = json.loads(line)
    if alert is None:
        continue
    if alert['healpix_url'] is None:
        continue
    if alert['dec'] <= 30.0 and alert['event_name'][0] not in LSST_names:
        LSST_alerts.append(alert)
        LSST_names.append(alert['event_name'][0])
    if alert['dec'] >= -30.0 and alert['event_name'][0] not in ZTF_names:
        ZTF_alerts.append(alert)
        ZTF_names.append(alert['event_name'][0])
    
    
    
    
```

```python
ZTF_last_50 =  ZTF_alerts[:-50]
LSST_last_50 =  LSST_alerts[:-50]
with open('LSST_simulated_alerts.txt', 'w') as file:
    json.dumps(LSST_alerts, file, indent=4)
with open('ZTF_simulated_alerts.txt', 'w') as file:
    json.dumps(ZTF_alerts, file, indent=4)
```
